# Live code

TeachBooks books can run Python code in the reader's browser. The extension saves the code a reader writes or changes in a code cell. When the reader opens the page again, their code is back in the cell.

## Try it

1. Click {fa}`rocket` and then **Live Code** at the top of this page. The code cells become editable.
2. Change the numbers in the first cell and run it.
3. Reload the page. Your changed code is back in the cell, even before you start Live Code again.
4. Run the cell again to see the output. The extension saves only the code, not the output.

```python
number_of_agents = 10
steps = 5

print(f"A model with {number_of_agents} agents runs for {steps} steps.")
```

```python
# Write your own code here and reload the page
total = 0
for step in range(steps):
    total = total + number_of_agents
print(total)
```

## How it works

Each saved edit belongs to a cell's place on the page. The extension also remembers what the original cell looked like. If you later change that cell in a new version of your book, the old edit no longer matches, and the extension skips it. So a reader never sees their code in a cell where it does not belong.

The live code part of TeachBooks asks the reader before they leave a page with unsaved code. The extension turns that question off, because the code is already saved.

Turn this part off with `persistence_code_cells: false`.
